### Heatmap correlation 

```python
import matplotlib.pyplot as plt
import seaborn as sns
cor=df.corr()

axis=sns.heatmap(cor,vmin=-1,vmax=1,center=0,square=True,cmap=sns.diverging_palette(3,300,n=500)


)
plt.show()
```
