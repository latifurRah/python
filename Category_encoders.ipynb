```python
import category_encoders as ce
import pandas as pd
train_df=pd.DataFrame({"Degree":["High School","Masters","Phd","Diploma","High School","Phd","Bachalors","Bachalors"]})
train_df
```

```python
encoder=ce.OrdinalEncoder(cols=['Degree'],return_df=True,mapping=[{"col":"Degree","mapping":{"None":0,"High School":1,
                                                                                             "Masters":2,"Phd":3,"Bachalors":4,
                                                                                              "Diploma":5}}])


train_df
```

```python
df=encoder.fit_transform(train_df)
df
```
